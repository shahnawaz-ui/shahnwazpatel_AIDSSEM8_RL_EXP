```python
print('shahnawaz patel / 232A012 / 32 / RL EXP 05 ')
import numpy as np
import random

#GridWorld Environment
class GridWorld:
    def __init__(self, size=4, start=(0, 0), goal=(3, 3)):
        self.size = size
        self.start = start
        self.goal = goal
        self.state = start
        self.actions = [(0, 1), (1, 0), (0, 1), (-1, 0)] # Right, Up, Down, Left, Up

    def reset(self):
        self.state = self.start
        return self.state

    def step(self, action):
        next_state = (self.state[0] + self.actions[action][0],
                      self.state[1] + self.actions[action][1])

        if 0 <= next_state[0] < self.size and 0 <= next_state[1] < self.size:
            self.state = next_state

        reward = 1 if self.state == self.goal else -0.1
        done = self.state == self.goal
        return self.state, reward, done
```

```python
print('shahnawaz patel / 232A012 / 32 / RL EXP 05 ')
import numpy as np
import random

def monte_carlo_control(env, episodes=5000, gamma=0.9, epsilon=0.1):
    Q = {((i, j), a): 0 for i in range(env.size) for j in range(env.size) for a in range(4)}
    # Initialize returns with lists, not tuples, to allow appending
    returns = {((i, j), a): [] for i in range(env.size) for j in range(env.size) for a in range(4)}

    for _ in range(episodes): # Added '_' as iteration variable
        state = env.reset()
        episode = [] # Stores (state, action, reward) tuples for the current episode
        visited_state_actions = set() # To track (state, action) pairs for first-visit MC

        while True:
            # Epsilon-greedy policy for action selection
            if random.random() < epsilon:
                action = random.choice(range(4)) # Explore: choose a random action
            else:
                # Exploit: choose the action with the highest Q-value for the current state
                # Use Q.get with a default value of -infinity for unvisited state-actions
                best_action = max(range(4), key=lambda a: Q.get((state, a), -float('inf')))
                action = best_action

            next_state, reward, done = env.step(action) # Added assignment operator
            episode.append((state, action, reward))
            state = next_state # Corrected variable name
            if done:
                break

        G = 0
        # Iterate through the episode in reverse to calculate returns
        for state_t, action_t, reward_t in reversed(episode):
            G = gamma * G + reward_t
            # First-visit Monte Carlo: only update if (state, action) pair is encountered for the first time in this episode segment
            if (state_t, action_t) not in visited_state_actions:
                returns[(state_t, action_t)].append(G)
                Q[(state_t, action_t)] = np.mean(returns[(state_t, action_t)])
                visited_state_actions.add((state_t, action_t))
    return Q
```

```python
print('shahnawaz patel / 232A012 / 32 / RL EXP 05 ')
# Temporal Difference Learning (SARSA)
def sarsa(env, episodes=5000, alpha=0.1, gamma=0.9, epsilon=0.1):
    Q = {((i, j), a): 0 for i in range(env.size) for j in range(env.size) for a in range(4)}
    for _ in range(episodes):
        state = env.reset()
        # Choose action A from state S using epsilon-greedy policy
        if random.random() < epsilon:
            action = random.choice(range(4))
        else:
            action = max(range(4), key=lambda a_val: Q.get((state, a_val), -float('inf')))

        while True:
            next_state, reward, done = env.step(action)
            # Choose next action A' from next_state S' using epsilon-greedy policy
            if random.random() < epsilon:
                next_action = random.choice(range(4))
            else:
                next_action = max(range(4), key=lambda a_val: Q.get((next_state, a_val), -float('inf')))

            # SARSA update
            Q[(state, action)] += alpha * (reward + gamma * Q.get((next_state, next_action), 0) - Q.get((state, action), 0))

            state, action = next_state, next_action
            if done:
                break
    return Q
```

```python
print('shahnawaz patel / 232A012 / 32/ RL EXP 05 ')
# Initialize Environment
env = GridWorld()
#Train with Monte Carlo
Q_mc = monte_carlo_control(env)
# Train with SARSA
Q_sarsa = sarsa(env)
# Train with Q-Learning
Q_qlearning = q_learning(env)
```
